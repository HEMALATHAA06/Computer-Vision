```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Input image path
image_path = r"C:\Users\shema\OneDrive\Desktop\image.png"
image = Image.open(image_path)

# Convert image to RGB numpy array
rgb_arr = np.array(image)

# Convert to CMY
cmy_arr = 255 - rgb_arr
cmy_image = Image.fromarray(cmy_arr.astype('uint8'), 'RGB')

# Save CMY image
cmy_path = r"C:\Users\shema\OneDrive\Desktop\image.png"
cmy_image.save(cmy_path)

# Split RGB channels
r_channel, g_channel, b_channel = image.split()

# Compute CMY channels
r_arr = np.array(r_channel)
g_arr = np.array(g_channel)
b_arr = np.array(b_channel)

c_channel = 255 - r_arr
m_channel = 255 - g_arr
y_channel = 255 - b_arr

# Convert CMY channels to images
c_image = Image.fromarray(c_channel.astype('uint8'), 'L')
m_image = Image.fromarray(m_channel.astype('uint8'), 'L')
y_image = Image.fromarray(y_channel.astype('uint8'), 'L')

# Save individual CMY channel images
c_path = r"C:\Users\shema\OneDrive\Desktop\image.png"
m_path = r"C:\Users\shema\OneDrive\Desktop\image.png"
y_path = r"C:\Users\shema\OneDrive\Desktop\image.png"
c_image.save(c_path)
m_image.save(m_path)
y_image.save(y_path)

# Plot results
plt.figure(figsize=(12, 4))
plt.subplot(1, 4, 1)
plt.imshow(cmy_image)
plt.title('CMY image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(c_channel, cmap='gray')
plt.title('Cyan Channel')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(m_channel, cmap='gray')
plt.title('Magenta Channel')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(y_channel, cmap='gray')
plt.title('Yellow Channel')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image using OpenCV
image2 = cv2.imread(r"C:\Users\shema\OneDrive\Desktop\image.png")

# Convert the BGR image to HSV
hsv_image1 = cv2.cvtColor(image2, cv2.COLOR_BGR2HSV)

# Split the HSV channels
h_channel1, s_channel1, v_channel1 = cv2.split(hsv_image1)

# Convert the HSV image to RGB for visualization
rgb_image = cv2.cvtColor(hsv_image1, cv2.COLOR_HSV2RGB)

# Plot the images
plt.figure(figsize=(15, 5))

# HSV Image (converted to RGB for display)
plt.subplot(1, 4, 1)
plt.title('HSV Image')
plt.imshow(rgb_image)
plt.axis('off')

# Hue Channel
plt.subplot(1, 4, 2)
plt.imshow(h_channel1, cmap='gray')
plt.title('Hue Channel')
plt.axis('off')

# Saturation Channel
plt.subplot(1, 4, 3)
plt.imshow(s_channel1, cmap='gray')
plt.title('Saturation Channel')
plt.axis('off')

# Value Channel
plt.subplot(1, 4, 4)
plt.imshow(v_channel1, cmap='gray')
plt.title('Value Channel')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image2=cv2.imread(r"C:\Users\shema\OneDrive\Desktop\image.png")
rgb_image = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
transformation_matrix = np.array([[0.299, 0.587, 0.114],
                                   [0.596, -0.274, -0.321],
                                   [0.211, -0.523, 0.311]])
reshaped_rgb = rgb_image.reshape((-1, 3))
yiq_image=np.dot(reshaped_rgb,transformation_matrix)
yiq_image = yiq_image.reshape(rgb_image.shape)
y_channel, i_channel, q_channel = cv2.split(yiq_image)

plt.figure(figsize=(15, 5))
plt.subplot(1, 4, 1)
plt.title('YIQ Image')
plt.imshow(yiq_image)
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(y_channel)  
plt.title('Y Channel')
plt.axis('off')
plt.subplot(1, 4, 3)
plt.imshow(i_channel)  
plt.title('I Channel')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(q_channel) 
plt.title('Q Channel')
plt.axis('off')
plt.show()
```
