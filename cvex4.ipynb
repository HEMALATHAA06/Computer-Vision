```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import seaborn as sns

# Load grayscale image
image = cv2.imread(r"C:\Users\shema\OneDrive\Desktop\jc.jpeg", 0)

hist, bins = np.histogram(image.ravel(), bins=50, range=[0, 256])
hist_right_shifted = np.roll(hist, 10)  
hist_left_shifted = np.roll(hist, -10)  
x_axis = bins[:-1]  
plt.figure(figsize=(12, 9))
plt.subplot(3, 2, 1)
plt.title("Original Image")
plt.imshow(image, cmap='gray')
plt.axis("off")
plt.subplot(3, 2, 2)
sns.histplot(image.ravel(), bins=50, kde=True, color='skyblue', edgecolor='black')
plt.title("Histogram (Original)")
plt.xlim(0, 255) 
plt.subplot(3, 2, 3)
plt.title("Original Image (Again)")
plt.imshow(image, cmap='gray')
plt.axis("off")
plt.subplot(3, 2, 4)
plt.bar(x_axis, hist_left_shifted, width=5, color='lightcoral', edgecolor='black')  
plt.title("left shift")
plt.xlim(0, 255) 
plt.subplot(3, 2, 5)
plt.title("Original Image (Again)")
plt.imshow(image, cmap='gray')
plt.axis("off")
plt.subplot(3, 2, 6)
plt.bar(x_axis, hist_right_shifted, width=5, color='skyblue', edgecolor='black')  
plt.title("Right shift")
plt.xlim(0, 255)  
plt.tight_layout()
plt.show()
```

```python
equalized_image = cv2.equalizeHist(image)
dft = cv2.dft(np.float32(image), flags=cv2.DFT_COMPLEX_OUTPUT)
dft_shift = np.fft.fftshift(dft)
magnitude_spectrum = cv2.magnitude(dft_shift[:, :, 0], dft_shift[:, :, 1])
dft_eq = cv2.dft(np.float32(equalized_image), flags=cv2.DFT_COMPLEX_OUTPUT)
dft_eq_shift = np.fft.fftshift(dft_eq)
magnitude_spectrum_eq = cv2.magnitude(dft_eq_shift[:, :, 0], dft_eq_shift[:, :, 1])
plt.figure(figsize=(12, 10))
plt.subplot(3, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')
plt.subplot(3, 2, 2)
sns.histplot(image.ravel(), bins=50, kde=True, color='skyblue', edgecolor='black')
plt.title("Histogram (Original)")
plt.subplot(3, 2, 3)
plt.imshow(equalized_image, cmap='gray')
plt.title('Histogram Equalized Image')
plt.axis('off')
plt.subplot(3, 2, 4)
sns.histplot(equalized_image.ravel(), bins=50, kde=True, color='lightcoral', edgecolor='black')
plt.title("Histogram (Equalized)")
plt.subplot(3, 2, 5)
plt.imshow(np.log(1 + magnitude_spectrum), cmap='gray')  
plt.title('Magnitude Spectrum (Original Image)')
plt.axis('off')
plt.subplot(3, 2, 6)
plt.imshow(np.log(1 + magnitude_spectrum_eq), cmap='gray')  
plt.title('Magnitude Spectrum (Equalized Image)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import seaborn as sns

# Load grayscale image
image = cv2.imread(r"C:\Users\shema\OneDrive\Desktop\jc.jpeg", 0)

# Compute histogram
hist, bins = np.histogram(image.ravel(), bins=50, range=[0, 256])

# Shift histogram
hist_right_shifted = np.roll(hist, 10)  
hist_left_shifted = np.roll(hist, -10)  
x_axis = bins[:-1]  
image_right_shifted = cv2.add(image, 50)  
image_left_shifted = cv2.subtract(image, 50)  
plt.figure(figsize=(12, 9))
plt.subplot(3, 2, 1)
plt.title("Original Image")
plt.imshow(image, cmap='gray')
plt.axis("off")
plt.subplot(3, 2, 2)
sns.histplot(image.ravel(), bins=50, kde=True, color='skyblue', edgecolor='black')
plt.title("Histogram (Original)")
plt.xlim(0, 255) 
plt.subplot(3, 2, 3)
plt.title("Right Shift")
plt.imshow(image_right_shifted, cmap='gray')
plt.axis("off")
plt.subplot(3, 2, 4)
plt.bar(x_axis, hist_right_shifted, width=5, color='lightblue', edgecolor='black')  
plt.title("Histogram (Right Shifted)")
plt.xlim(0, 255) 
plt.subplot(3, 2, 5)
plt.title("Left Shift")
plt.imshow(image_left_shifted, cmap='gray')
plt.axis("off")
plt.subplot(3, 2, 6)
plt.bar(x_axis, hist_left_shifted, width=5, color='lightcoral', edgecolor='black')  
plt.title("Histogram (Left Shifted)")
plt.xlim(0, 255)  
plt.tight_layout()
plt.show()
```
