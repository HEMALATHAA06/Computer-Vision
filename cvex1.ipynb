```python
import cv2
ig = cv2.imread(r"C:\Users\shema\OneDrive\Desktop\wp14908058.webp") 
ig 
import matplotlib.pyplot as plt 
 
def rgbtogray(originalig): 
    r, g, b = originalig[:, :, 0], originalig[:, :, 1], originalig[:, :, 2] 
    grayig = (0.21 * r) + (0.72 * g) + (0.07 * b) 
    return grayig 
 
def showig(originalig, grayig): 
    plt.subplot(1, 2, 1) 
    plt.imshow(cv2.cvtColor(originalig, cv2.COLOR_BGR2RGB))   
    plt.title("Original Image") 
 
    plt.subplot(1, 2, 2) 
    plt.imshow(grayig, cmap='gray') 
    plt.title("GrayScale Image") 
 
    plt.show() 
    
g2 = rgbtogray(ig) 
showig(ig, g2) 
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image_path = r"C:\Users\shema\OneDrive\Desktop\wp14908058.webp"
gray_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
equalized_image = cv2.equalizeHist(gray_image)
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(equalized_image, cmap='gray')
plt.title("Equalized Grayscale Image")
plt.axis("off")
plt.tight_layout()
plt.show()
def plot_histogram(image, title, color):
    hist, bins = np.histogram(image.flatten(), 256, [0, 256])
    plt.bar(range(256), hist, color=color, width=1.0)
    plt.title(title)
    plt.xlabel("Pixel Value")
    plt.ylabel("Frequency")
    plt.xlim([0, 256])
    plt.show()
plot_histogram(gray_image, "Original Grayscale Histogram", color='blue')
plot_histogram(equalized_image, "Equalized Grayscale Histogram", color='red')
```
