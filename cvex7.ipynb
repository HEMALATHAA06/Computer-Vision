```python
import matplotlib.pyplot as plt
import numpy as np
import cv2

image = cv2.imread(r'hars.png', 0)

plt.imshow(image, cmap='gray')
plt.axis("off")
```

```python
dst = cv2.cornerHarris(image, 2, 3, 0.04)
dst = cv2.dilate(dst,None)
plt.imshow(dst, cmap='gray')
plt.axis("off")
```

```python
thresh = 0.1*dst.max()

corner_image = np.copy(image)

for j in range(0, dst.shape[0]):
    for i in range(0, dst.shape[1]):
        if(dst[j,i] > thresh):

            cv2.circle( corner_image, (i, j), 1, (0,255,0), 1)

plt.imshow(corner_image, cmap='gray')
plt.axis("off")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread('hars.png')
operatedImage = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
operatedImage = np.float32(operatedImage)
dest = cv2.cornerHarris(operatedImage, 2, 5, 0.07)
dest = cv2.dilate(dest, None)
image[dest > 0.01 * dest.max()]=[0, 0, 255]
plt.imshow(image, cmap="gray")
plt.axis("off")
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('tile.jpg',cv2.IMREAD_COLOR)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray, 50, 200)

lines = cv2.HoughLinesP(edges, 1, np.pi/360, 360, minLineLength=15, maxLineGap=250)

for line in lines:
   x1, y1, x2, y2 = line[0]
   cv2.line(img, (x1, y1), (x2, y2), (255, 0, 0), 3)

print("Line Detection using Hough Transform")
plt.imshow(img)
plt.axis("off")
```
