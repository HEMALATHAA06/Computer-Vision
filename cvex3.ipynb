```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\shema\OneDrive\Desktop\seoul2.webp") 

rows, cols, channels = img.shape  
tx, ty = 150, 150 
translated_img = np.zeros_like(img)

for x in range(rows):
    for y in range(cols):
        new_x, new_y = x + tx, y + ty
        if 0 <= new_x < rows and 0 <= new_y < cols:
            translated_img[new_x, new_y] = img[x, y]  

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
translated_rgb = cv2.cvtColor(translated_img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img_rgb)

plt.subplot(1, 2, 2)
plt.title("Translated Image")
plt.imshow(translated_rgb)

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\shema\OneDrive\Desktop\seoul2.webp") 
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

rows, cols, channels = img.shape 
scaled_img=np.zeros_like(img_rgb)
sx=0.7
sy=1.0
for i in range(rows):
    for j in range(cols):
       xn=int(i * sx)
       yn=int(j * sy)
     
       if 0 <= xn < rows and 0 <= yn < cols:
            scaled_img[xn,yn]=img_rgb[i,j]

plt.subplot(1,2,1)
plt.title("Original image")
plt.imshow(img_rgb)

plt.subplot(1,2,2)
plt.title("Scaled image")
plt.imshow(scaled_img)
```

```python
from math import radians
angle=radians(30)
rotate=np.zeros_like(img_rgb)
rows, cols, channels = img_rgb.shape
xc=rows//2
yc=cols//2
for i in range(rows):
    for j in range(cols):
       x=i-xc
       y=j-yc
       xn=int((x * np.cos(angle)) -(y * np.sin(angle))) + xc
       yn=int((x * np.sin(angle)) + (y * np.cos(angle))) + yc
       if 0 <= xn < rows and 0 <= yn < cols:
            rotate[xn,yn]=img_rgb[i,j]

plt.subplot(1,2,1)
plt.title("Original image")
plt.imshow(img_rgb)

plt.subplot(1,2,2)
plt.title("Rotated image")
plt.imshow(rotate)

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image
img = cv2.imread(r"C:\Users\shema\OneDrive\Desktop\seoul2.webp")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

rows, cols, channels = img_rgb.shape

sx = 0.3  
sy = 0.3  

shear_img_both = np.zeros((int(rows + sy * cols), int(cols + sx * rows), channels), dtype=np.uint8)

for i in range(rows):
    for j in range(cols):
        new_x = int(j + sx * i)  
        new_y = int(i + sy * j)
        if 0 <= new_x < shear_img_both.shape[1] and 0 <= new_y < shear_img_both.shape[0]:
            shear_img_both[new_y, new_x] = img_rgb[i, j]

# Plotting the images
plt.figure(figsize=(10, 7))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img_rgb)
plt.axis("off")

plt.subplot(1, 2, 2)
plt.title("Shear Image (both-axis)")
plt.imshow(shear_img_both)
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image
img = cv2.imread(r"C:\Users\shema\OneDrive\Desktop\seoul2.webp")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Get image dimensions
rows, cols, channels = img_rgb.shape

# Shear factors
sx = 0.3 # Shear factor along x-axis
sy = 0.3 # Shear factor along y-axis

# Create an empty image with the same size as the original
shear_img_both = np.zeros_like(img_rgb)

for i in range(rows):
    for j in range(cols):
        new_x = int(j + sx * i)  # Shearing along x-axis
        new_y = int(i + sy * j)  # Shearing along y-axis
        
        # Ensure the transformed coordinates remain within image bounds
        if 0 <= new_x < cols and 0 <= new_y < rows:
            shear_img_both[new_y, new_x] = img_rgb[i, j]

# Plot the images
plt.figure(figsize=(10, 7))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(img_rgb)
plt.axis("off")

plt.subplot(1, 2, 2)
plt.title("Shear Image (both-axis)")
plt.imshow(shear_img_both)
plt.axis("off")

plt.show()
```
